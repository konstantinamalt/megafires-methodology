```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import shape
import json
```

```python
df = gpd.read_file("forest_fires_nutsall_mapbox_4326.geojson")
```

```python
winter_season = ['January', 'February', 'March', 'November', 'December']
fire_season = ['April', 'May', 'June', 'July', 'August', 'September', 'October']

df_fire_season = df[df['month'].isin(fire_season)]
df_winter_season = df[df['month'].isin(winter_season)]
```

```python
natura = gpd.read_file(
    r"C:\Users\kwnst\Downloads\forest_fires\SHP\SHP\Natura2000_end2024_epsg3035.shp"
)
df_3035 = df_fire_season.to_crs(natura.crs)
```

```python
overlaps = gpd.overlay(
    df_3035,
    natura,
    how="intersection",
    keep_geom_type=True
)
overlaps
```

```python
overlaps.columns.tolist()
```

```python
overlaps["overlap_natura2k_ha"] = overlaps.geometry.area / 10_000
```

```python
overlaps["official_link"] = (
    "https://natura2000.eea.europa.eu/Natura2000/SDF.aspx?site="
    + overlaps["SITECODE"]
)
```

```python
overlaps_large = overlaps[
    overlaps["area_ha"] >= 5000
].copy()
```

```python
repeated_sites = (
    overlaps_large
    .groupby(["SITECODE", "SITENAME"], dropna=False)
    .agg(
        fire_count=("id", "nunique"),
        fire_ids=("id", lambda x: sorted(x.dropna().unique())),
        years=("year", lambda x: sorted(x.dropna().unique())),
        months=("month", lambda x: sorted(x.dropna().unique())),
        years_affected=("year", "nunique"),
        countries=("country", lambda x: sorted(x.dropna().unique())),
        nuts3_names=("nuts3_name", lambda x: sorted(x.dropna().unique())),
        provinces=("province", lambda x: sorted(x.dropna().unique())),
        total_overlap_ha=("overlap_natura2k_ha", "sum")
    )
    .reset_index()
)

repeated_sites = (
    repeated_sites[repeated_sites["fire_count"] > 1]
    .sort_values(["years_affected", "total_overlap_ha"], ascending=False)
)

repeated_sites["official_link"] = (
    "https://natura2000.eea.europa.eu/Natura2000/SDF.aspx?site="
    + repeated_sites["SITECODE"]
)
repeated_sites.to_excel("repeated_sites_natura2k_decade_fire_season.xlsx", index=False)
repeated_sites
```

```python
df_3035_all = df.to_crs(natura.crs)
overlaps_all = gpd.overlay(
    df_3035_all,
    natura,
    how="intersection",
    keep_geom_type=True
)
overlaps_all["overlap_natura2k_ha"] = overlaps.geometry.area / 10_000
```

```python
overlaps_medium_large = overlaps_all[
    overlaps_all["area_ha"] >= 1000
].copy()

repeated_sites2 = (
    overlaps_medium_large
    .groupby(["SITECODE", "SITENAME"], dropna=False)
    .agg(
        fire_count=("id", "nunique"),
        fire_ids=("id", lambda x: sorted(x.dropna().unique())),
        years=("year", lambda x: sorted(x.dropna().unique())),
        months=("month", lambda x: sorted(x.dropna().unique())),
        years_affected=("year", "nunique"),
        countries=("country", lambda x: sorted(x.dropna().unique())),
        nuts3_names=("nuts3_name", lambda x: sorted(x.dropna().unique())),
        provinces=("province", lambda x: sorted(x.dropna().unique())),
        total_overlap_ha=("overlap_natura2k_ha", "sum")
    )
    .reset_index()
)

repeated_sites2 = (
    repeated_sites2[repeated_sites2["fire_count"] > 1]
    .sort_values(["years_affected", "total_overlap_ha"], ascending=False)
)

repeated_sites2 = repeated_sites2.round(2)
repeated_sites2.to_excel("repeated_sites_above1k_ha_ALL.xlsx", index=False)

repeated_sites2
```

```python
repeated_sites_above1k_ha_greece = repeated_sites2[
    repeated_sites2['countries'].apply(lambda x: 'EL' in x if isinstance(x, list) else False)
]
repeated_sites_above1k_ha_greece.to_excel("repeated_sites_above1k_ha_greece_fire_season.xlsx", index=False)
repeated_sites_above1k_ha_greece
```
