```python
# Define fire size categories
bins = [0, 5, 30, 100, 1000, 5000, 10000, float("inf")]

labels = [
    "Very small (<5 ha)",
    "Small (5-30 ha)", #until <30 and then >=30 in the next category
    "Medium (30-100 ha)",
    "Large (100-1,000 ha)",
    "Very large (1,000-5,000 ha)",
    "Extreme (5,000-10,000 ha)",
    "Mega (10,000+ ha)"
]

df["fire_size_category"] = pd.cut(
    df["area_ha"],
    bins=bins,
    labels=labels,
    right=False,
    include_lowest=True
)

df.drop(columns=['nuts_country', 'geometry_area_ha', 'total_geometry_area_ha'], inplace=True)
df
```

```python
master_yearly = (
    italy.groupby(["country", "year"])["id"]
    .nunique()
    .reset_index(name="Number of fires")
)

master_yearly
```

```python
yearly_area = (
    italy.groupby(["country", "year"])["area_ha"]
    .sum()
    .reset_index(name="Burned hectares per year")
)
yearly_area
```

```python
master_y_ha = master_yearly.merge(
    yearly_area,
    on=["country", "year"],
    how="left"
)

master_y_ha
```

```python
yearly_land_cover = (
    italy.groupby(["country", "year"])[
        [
            "natural_vegetation_ha",
            "forests_ha",
            "agricultural_areas_ha",
            "artificial_surfaces_ha",
            "other_land_cover_ha",
            "natura2000_ha"
        ]
    ]
    .sum(min_count=1)
    .reset_index()
)

yearly_land_cover
```

```python
master_italy = master_y_ha.merge(
    yearly_land_cover,
    on=["country", "year"],
    how="left"
)

master_italy
```

```python
master_eu_yearly = (
    unique_fires
    .groupby("year")
    .agg(
        **{
            "EU Number of fires": ("id", "nunique"),
            "EU Burned hectares": ("area_ha", "sum"),
            "EU Natural vegetation burned (ha)": (
                "natural_vegetation_ha",
                lambda x: x.sum(min_count=1)
            ),
            "EU Forests burned (ha)": (
                "forests_ha",
                lambda x: x.sum(min_count=1)
            ),
            "EU Agricultural areas burned (ha)": (
                "agricultural_areas_ha",
                lambda x: x.sum(min_count=1)
            ),
            "EU Artificial surfaces burned (ha)": (
                "artificial_surfaces_ha",
                lambda x: x.sum(min_count=1)
            ),
            "EU Other land cover burned (ha)": (
                "other_land_cover_ha",
                lambda x: x.sum(min_count=1)
            ),
            "EU Natura 2000 burned (ha)": (
                "natura2000_ha",
                lambda x: x.sum(min_count=1)
            )
        }
    )
    .reset_index()
    .rename(columns={"year": "EU Year"})
    .sort_values("EU Year")
)

master_eu_yearly
```

```python
master_italy = (
    master_italy
    .merge(
        master_eu_yearly,
        left_on="year",
        right_on="EU Year",
        how="left"
    )
    .drop(columns="EU Year")
)

master_italy
```

```python
column_order = [
    "country",
    "year",

    "Number of fires",
    "EU Number of fires",

    "Burned hectares per year",
    "EU Burned hectares",

    "natural_vegetation_ha",
    "EU Natural vegetation burned (ha)",

    "forests_ha",
    "EU Forests burned (ha)",

    "agricultural_areas_ha",
    "EU Agricultural areas burned (ha)",

    "artificial_surfaces_ha",
    "EU Artificial surfaces burned (ha)",

    "other_land_cover_ha",
    "EU Other land cover burned (ha)",

    "natura2000_ha",
    "EU Natura 2000 burned (ha)"
]

master_italy = master_italy[column_order]

master_italy
```

```python
master_italy["Italy share of EU fires (%)"] = (
    master_italy["Number of fires"]
    / master_italy["EU Number of fires"]
    * 100
).round(2)

master_italy["Italy share of EU burned hectares (%)"] = (
    master_italy["Burned hectares per year"]
    / master_italy["EU Burned hectares"]
    * 100
).round(2)

master_italy["Italy share of EU natural vegetation burned (%)"] = (
    master_italy["natural_vegetation_ha"]
    / master_italy["EU Natural vegetation burned (ha)"]
    * 100
).round(2)

master_italy["Italy share of EU forests burned (%)"] = (
    master_italy["forests_ha"]
    / master_italy["EU Forests burned (ha)"]
    * 100
).round(2)

master_italy["Italy share of EU agricultural areas burned (%)"] = (
    master_italy["agricultural_areas_ha"]
    / master_italy["EU Agricultural areas burned (ha)"]
    * 100
).round(2)

master_italy["Italy share of EU artificial surfaces burned (%)"] = (
    master_italy["artificial_surfaces_ha"]
    / master_italy["EU Artificial surfaces burned (ha)"]
    * 100
).round(2)

master_italy["Italy share of EU other land cover burned (%)"] = (
    master_italy["other_land_cover_ha"]
    / master_italy["EU Other land cover burned (ha)"]
    * 100
).round(2)

master_italy["Italy share of EU Natura 2000 burned (%)"] = (
    master_italy["natura2000_ha"]
    / master_italy["EU Natura 2000 burned (ha)"]
    * 100
).round(2)

master_italy
```

```python
column_order = [
    "country",
    "year",

    "Number of fires",
    "EU Number of fires",
    "Italy share of EU fires (%)",

    "Burned hectares per year",
    "EU Burned hectares",
    "Italy share of EU burned hectares (%)",

    "natural_vegetation_ha",
    "EU Natural vegetation burned (ha)",
    "Italy share of EU natural vegetation burned (%)",

    "forests_ha",
    "EU Forests burned (ha)",
    "Italy share of EU forests burned (%)",

    "agricultural_areas_ha",
    "EU Agricultural areas burned (ha)",
    "Italy share of EU agricultural areas burned (%)",

    "artificial_surfaces_ha",
    "EU Artificial surfaces burned (ha)",
    "Italy share of EU artificial surfaces burned (%)",

    "other_land_cover_ha",
    "EU Other land cover burned (ha)",
    "Italy share of EU other land cover burned (%)",

    "natura2000_ha",
    "EU Natura 2000 burned (ha)",
    "Italy share of EU Natura 2000 burned (%)"
]

master_italy = master_italy[column_order]
master_italy
```

```python
master_italy["Fires (% of Italy total)"] = (
    master_italy["Number of fires"]
    / master_italy["Number of fires"].sum()
    * 100
).round(2)

master_italy["Burned hectares (% of Italy total)"] = (
    master_italy["Burned hectares per year"]
    / master_italy["Burned hectares per year"].sum()
    * 100
).round(2)

master_italy["Natural vegetation (% of Italy total)"] = (
    master_italy["natural_vegetation_ha"]
    / master_italy["natural_vegetation_ha"].sum()
    * 100
).round(2)

master_italy["Forests (% of Italy total)"] = (
    master_italy["forests_ha"]
    / master_italy["forests_ha"].sum()
    * 100
).round(2)

master_italy["Agricultural areas (% of Italy total)"] = (
    master_italy["agricultural_areas_ha"]
    / master_italy["agricultural_areas_ha"].sum()
    * 100
).round(2)

master_italy["Artificial surfaces (% of Italy total)"] = (
    master_italy["artificial_surfaces_ha"]
    / master_italy["artificial_surfaces_ha"].sum()
    * 100
).round(2)

master_italy["Other land cover (% of Italy total)"] = (
    master_italy["other_land_cover_ha"]
    / master_italy["other_land_cover_ha"].sum()
    * 100
).round(2)

master_italy["Natura 2000 (% of Italy total)"] = (
    master_italy["natura2000_ha"]
    / master_italy["natura2000_ha"].sum()
    * 100
).round(2)

master_italy
```

```python
column_order = [
    "country",
    "year",

    # Fires
    "Number of fires",
    "Fires (% of Italy total)",
    "EU Number of fires",
    "Italy share of EU fires (%)",

    # Total burned area
    "Burned hectares per year",
    "Burned hectares (% of Italy total)",
    "EU Burned hectares",
    "Italy share of EU burned hectares (%)",

    # Natural vegetation
    "natural_vegetation_ha",
    "Natural vegetation (% of Italy total)",
    "EU Natural vegetation burned (ha)",
    "Italy share of EU natural vegetation burned (%)",

    # Forests
    "forests_ha",
    "Forests (% of Italy total)",
    "EU Forests burned (ha)",
    "Italy share of EU forests burned (%)",

    # Agricultural areas
    "agricultural_areas_ha",
    "Agricultural areas (% of Italy total)",
    "EU Agricultural areas burned (ha)",
    "Italy share of EU agricultural areas burned (%)",

    # Artificial surfaces
    "artificial_surfaces_ha",
    "Artificial surfaces (% of Italy total)",
    "EU Artificial surfaces burned (ha)",
    "Italy share of EU artificial surfaces burned (%)",

    # Other land cover
    "other_land_cover_ha",
    "Other land cover (% of Italy total)",
    "EU Other land cover burned (ha)",
    "Italy share of EU other land cover burned (%)",

    # Natura 2000
    "natura2000_ha",
    "Natura 2000 (% of Italy total)",
    "EU Natura 2000 burned (ha)",
    "Italy share of EU Natura 2000 burned (%)"
]

master_italy = master_italy[column_order]
master_italy
```
